```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data with correct header
df = pd.read_csv(r"C:\Users\yamam\OneDrive\Desktop\sales_data.csv", header=1)
print(df.columns.tolist())
print(df.head())

# Summary
summary = df.groupby("CITY")["SALES"].sum().reset_index()
print(summary)

# Chart
plt.figure(figsize=(8,5))
plt.bar(summary["CITY"], summary["SALES"], color=["#4CAF50","#2196F3","#FF9800","#E91E63"])
plt.title("Total Sales by City")
plt.xlabel("City")
plt.ylabel("Sales")
plt.tight_layout()
plt.savefig(r"C:\Users\yamam\OneDrive\Desktop\sales_by_city.png")
plt.show()
print("Chart saved to Desktop!")
```

```python
# Chart 2 - Sales pie chart
plt.figure(figsize=(7,7))
plt.pie(summary["SALES"], labels=summary["CITY"], autopct="%1.1f%%",
        colors=["#4CAF50","#2196F3","#FF9800","#E91E63"])
plt.title("Sales Distribution by City")
plt.savefig(r"C:\Users\yamam\OneDrive\Desktop\sales_pie.png")
plt.show()
print("Pie chart saved!")
```
